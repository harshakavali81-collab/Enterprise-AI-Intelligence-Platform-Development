### Hyderabad Sales Trend & Regional Breakdown
Analyzing the month-over-month revenue dip in the Hyderabad branch.

```python
sales['month'] = pd.to_datetime(sales['sale_date']).dt.to_period('M')
monthly_city = sales.groupby(['month', 'city'])['revenue'].sum().unstack()
print('Monthly Revenue by City (Last 3 Months):')
display(monthly_city.tail(3))
```

# 02. Feature Engineering Pipeline
## Customer Churn & Demand Forecasting Feature Stores

Key transformations:
- **RFM Metrics**: Recency (`days_since_last_order`), Frequency (`order_frequency`), Monetary (`total_spend`).
- **Engagement Ratios**: Support tickets per frequency ratio, Average Order Value (AOV).
- **One-Hot & Categorical Encodings**: Enterprise vs SMB segment flags.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/sample/customers.csv')
df['aov'] = df['total_spend'] / df['order_frequency'].clip(lower=1)
df['ticket_ratio'] = df['support_tickets_count'] / df['order_frequency'].clip(lower=1)
df['is_enterprise'] = (df['segment'] == 'Enterprise').astype(int)

feature_cols = ['days_since_last_order', 'order_frequency', 'total_spend', 'support_tickets_count', 'is_enterprise', 'aov']
X = df[feature_cols]
y = df['is_churned']
print('Engineered features summary:')
display(X.describe())
```
